###Phase 1 – Data Understanding & EDA

```python
# Show all columns in output
pd.set_option('display.max_columns', None)
pd.set_option('display.width', 200)
```

```python
# -------------------------------
# 1. Shape & Info
# -------------------------------
print("\n📦 Dataset Shape:")
print(df.shape)

print("\n🔍 Data Types and Non-Null Counts:")
print(df.info())
#no nulls in all the 41 columns 
```

```python
# 2. Missing Values
# -------------------------------
print("\n❓ Missing Values per Column:")
missing = df.isnull().sum()
print(missing[missing >= 0].sort_values(ascending=False))
```

```python
# -------------------------------
# 3. Unique Values
# -------------------------------
print("\n🔁 Unique Values per Column:")
unique_counts = df.nunique().sort_values(ascending=False)
print(unique_counts)
```

```python
# -------------------------------
# 4. Summary Statistics
# -------------------------------
print("\n📊 Summary Statistics (Numerical):")
print(df.describe())
```

```python
print("\n📊 Summary Statistics (Categorical):")
print(df.describe(include='object'))
```

```python
# -------------------------------
# 5. Correlation Heatmap
# -------------------------------
print("\n Correlation Matrix (Numerical Features):")
num_df = df.select_dtypes(include=[np.number])
if not num_df.empty:
    plt.figure(figsize=(10, 6))
    sns.heatmap(num_df.corr(), cmap='coolwarm')
    plt.title('Correlation Heatmap')
    plt.tight_layout()
    plt.show()
else:
    print("No numeric columns found.")
```

```python
# 6. Univariate Distribution Plots
# -------------------------------
print("\n📉 Univariate Distribution Plots:")

# Numerical features
for col in num_df.columns:
    plt.figure(figsize=(6, 4))
    sns.histplot(df[col], kde=True)
    plt.title(f'Distribution of {col}')
    plt.tight_layout()
    plt.show()
```

```python
# Categorical features
cat_df = df.select_dtypes(include='object')
for col in cat_df.columns:
    plt.figure(figsize=(8, 4))
    sns.countplot(data=df, x=col, order=df[col].value_counts().index)
    plt.title(f'Frequency of {col}')
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()
```

```python
target = "label"

# 1. Select numerical columns
numerical_cols = df.select_dtypes(include=['int64', 'float64']).columns.tolist()

# 2. Add target column if not already in numerical
if target not in numerical_cols:
    numerical_cols.append(target)

# 3. Compute correlation matrix
correlation_matrix = df[numerical_cols].corr()

# 4. Show correlation of all numerical features with the target
correlation_with_target = correlation_matrix[target].sort_values(ascending=False)

print("Correlation with Target:\n", correlation_with_target)
```

```python
# Heatmap of correlations
plt.figure(figsize=(10, 4))
sns.heatmap(df[numerical_cols].corr(), annot=True, cmap='coolwarm', fmt=".2f")
plt.title('Correlation Matrix (Numerical Features)')
plt.show()
```

```python
# Define your target label
target = 'label'  # Change this to your actual target column

# Select only numerical features (excluding the target)
numerical_cols = df.select_dtypes(include=['number']).columns.tolist()
if target not in numerical_cols:
    print(f"'{target}' is not numeric. You may need to encode it first.")
else:
    # Drop target from feature list
    numerical_cols.remove(target)

    # Compute correlation with target only
    target_corr = df[numerical_cols + [target]].corr()[[target]].drop(index=target)

    # Plot heatmap
    plt.figure(figsize=(6, len(numerical_cols) * 0.5))
    sns.heatmap(target_corr, annot=True, cmap='coolwarm', fmt=".2f")
    plt.title(f'Correlation of Numerical Features with "{target}"')
    plt.tight_layout()
    plt.show()
```

```python
# 1. Select categorical columns
categorical_cols = df.select_dtypes(include=['object', 'category']).columns.tolist()

# 2. Plot target distribution per category
for col in categorical_cols:
    plt.figure(figsize=(10, 4))
    sns.boxplot(x=col, y=target, data=df)
    plt.title(f'{target} by {col}')
    plt.xticks(rotation=45)
    plt.show()
```

```python
# Encode categorical variables using LabelEncoder
from sklearn.preprocessing import LabelEncoder

df_encoded = df.copy()
label_encoders = {}

for col in categorical_cols:
    le = LabelEncoder()
    df_encoded[col] = le.fit_transform(df[col].astype(str))
    label_encoders[col] = le  # in case you need to inverse later

# Now compute correlations with target
encoded_corrs = df_encoded[categorical_cols + [target]].corr()[target].sort_values(ascending=False)
print("Encoded Categorical Feature Correlations with Target:\n", encoded_corrs)
```
