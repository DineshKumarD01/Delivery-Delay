We'll build from the **ground up**, with **clear definitions**, **step-by-step derivations**, and **decision criteria** for selecting the right metric depending on your **ML/DS problem context**.

---

## 🔷 I. Multi-Class Classification Basics

In **multi-class classification**, we predict one of **$K$** classes (e.g., predicting digit 0–9, disease type, animal type, etc.).

For evaluation:

* Use a **confusion matrix** for **each class**: treat the current class as "positive" and all others as "negative" — known as **one-vs-rest (OvR)** or **one-vs-all** strategy.

---

## 🔷 II. Confusion Matrix for Each Class

For **each class $c \in \{1, 2, ..., K\}$**:

|            | Predicted = c | Predicted ≠ c |
| ---------- | ------------- | ------------- |
| Actual = c | **TP**        | **FN**        |
| Actual ≠ c | **FP**        | **TN**        |

You compute metrics for **each class**, then average them across classes.

---

## 🔷 III. Key Evaluation Metrics (Formulas + Intuition)

### ✅ 1. **Accuracy**

**Definition**: Fraction of correct predictions among total predictions.

**Formula**:

$$
\text{Accuracy} = \frac{\text{Number of correct predictions}}{\text{Total number of samples}} = \frac{\sum_{i=1}^{K} \text{TP}_i}{N}
$$

Where:

* $TP_i$ = true positives for class $i$
* $N$ = total number of instances

**Pros**:

* Easy to interpret.
* Useful when class distribution is **balanced**.

**Cons**:

* Misleading with **imbalanced classes**.

---

### ✅ 2. **Precision (per class)**

**Definition**: Among all predicted as class $c$, how many are truly class $c$?

$$
\text{Precision}_c = \frac{TP_c}{TP_c + FP_c}
$$

**Intuition**:

* High precision = **low false positives**.
* Important when **false positives are costly** (e.g., predicting disease when patient is healthy).

---

### ✅ 3. **Recall (Sensitivity / True Positive Rate)**

**Definition**: Out of all actual samples of class $c$, how many did we correctly predict?

$$
\text{Recall}_c = \frac{TP_c}{TP_c + FN_c}
$$

**Intuition**:

* High recall = **low false negatives**.
* Crucial when **false negatives are costly** (e.g., cancer detection).

---

### ✅ 4. **F1 Score (per class)**

**Definition**: Harmonic mean of precision and recall.

$$
\text{F1}_c = 2 \cdot \frac{\text{Precision}_c \cdot \text{Recall}_c}{\text{Precision}_c + \text{Recall}_c}
$$

**Why harmonic mean?**

* It penalizes **imbalanced precision/recall**.
* You get a high F1 **only** when both precision and recall are high.

---

### ✅ 5. **Macro, Micro, and Weighted Averaging**

Since we have **multiple classes**, we aggregate class-wise metrics.

#### A. Macro Averaging

**Equal weight to each class.**

$$
\text{Macro-F1} = \frac{1}{K} \sum_{i=1}^{K} \text{F1}_i
$$

✅ Best when you want to **treat all classes equally**, **even rare ones**.

#### B. Micro Averaging

**Aggregate all TP, FP, FN before computing metrics.**

$$
\text{Micro-Precision} = \frac{\sum_{i} TP_i}{\sum_i (TP_i + FP_i)}
$$

$$
\text{Micro-Recall} = \frac{\sum_{i} TP_i}{\sum_i (TP_i + FN_i)}
$$

$$
\text{Micro-F1} = \frac{2 \cdot \text{Micro-Precision} \cdot \text{Micro-Recall}}{\text{Micro-Precision} + \text{Micro-Recall}}
$$

✅ Best when classes are **imbalanced** and you want to evaluate **overall performance**.

#### C. Weighted Averaging

**Weighted by class support (number of true instances per class).**

$$
\text{Weighted-F1} = \sum_{i=1}^{K} \left( \frac{n_i}{N} \cdot \text{F1}_i \right)
$$

✅ Good for **imbalanced datasets**, gives more importance to frequent classes.

---

### ✅ 6. **Cohen’s Kappa**

Measures agreement between predicted and actual labels **adjusted for chance**.

$$
\kappa = \frac{p_o - p_e}{1 - p_e}
$$

Where:

* $p_o$ = observed accuracy
* $p_e$ = expected accuracy by chance

✅ Better than accuracy when class distribution is skewed.

---

### ✅ 7. **Matthews Correlation Coefficient (MCC)**

Generalized to multi-class:

$$
\text{MCC} = \frac{c \cdot s - \sum_k p_k \cdot t_k}{\sqrt{(c^2 - \sum_k p_k^2) \cdot (c^2 - \sum_k t_k^2)}}
$$

Where:

* $c = \sum_k CM_{kk}$ = total correct predictions
* $s = \sum_{i,j} CM_{ij}$ = total samples
* $p_k = \sum_i CM_{ik}$ = predicted as class k
* $t_k = \sum_j CM_{kj}$ = actual class k

✅ Excellent for **imbalanced** multi-class classification.
✅ Produces a score between -1 and 1:

* 1 = perfect prediction
* 0 = random
* -1 = total disagreement

---

### ✅ 8. **Log Loss / Cross Entropy Loss**

For probabilistic classifiers (like softmax):

$$
\text{LogLoss} = -\frac{1}{N} \sum_{i=1}^{N} \sum_{j=1}^{K} y_{ij} \cdot \log(\hat{y}_{ij})
$$

Where:

* $y_{ij}$ = 1 if instance $i$ belongs to class $j$, else 0
* $\hat{y}_{ij}$ = predicted probability of instance $i$ for class $j$

✅ Penalizes incorrect predictions **with high confidence**.
✅ Best for probabilistic models and **model selection**.

---

## 🔷 IV. When to Use Which Metric?

| Scenario                                     | Preferred Metric(s)                       | Why                                    |
| -------------------------------------------- | ----------------------------------------- | -------------------------------------- |
| Balanced classes                             | Accuracy, Micro-F1                        | Straightforward, works well            |
| Imbalanced classes                           | Macro-F1, Weighted-F1, MCC, Cohen’s Kappa | Macro/Weighted treat classes fairly    |
| Want to penalize wrong confident predictions | Log Loss                                  | Takes probability into account         |
| Optimizing for rare class (e.g., fraud)      | Recall (macro), Macro-F1                  | Don’t miss positives                   |
| Real-time system (false positives dangerous) | Precision                                 | Avoid false alarms                     |
| Overall consistency                          | MCC                                       | Balanced, interpretable even if skewed |

## 🔷 V. Example to Reinforce Intuition

Let’s say you have 3 classes: **A, B, C** and 100 samples.

Confusion matrix (true vs predicted):

|        | Pred A | Pred B | Pred C |
| ------ | ------ | ------ | ------ |
| True A | 30     | 5      | 5      |
| True B | 2      | 25     | 3      |
| True C | 1      | 2      | 27     |

### Class-wise Metrics:

For class A:

* TP = 30
* FP = 2 (from B) + 1 (from C) = 3
* FN = 5 + 5 = 10
* Precision = 30 / (30 + 3) = 0.909
* Recall = 30 / (30 + 10) = 0.75
* F1 = 2 \* (0.909 \* 0.75) / (0.909 + 0.75) ≈ 0.822

Repeat for B, C.

Then compute:

* Macro-F1 = average of F1\_A, F1\_B, F1\_C
* Micro-F1 = total TP / total (TP + FP + FN)
* Accuracy = (30 + 25 + 27) / 100 = 82%
* Log Loss = computed using predicted probabilities.

Great follow-up, Dragon. You're absolutely right to ask about **evaluation curves** in multi-class classification — they provide **threshold-independent** and **probability-aware** views of model performance.

Let’s dive **step-by-step** into:

---

## 🔷 I. Types of Evaluation Curves

For **multi-class classification**, the two most common curves are:

1. **ROC Curve (Receiver Operating Characteristic)**
2. **Precision-Recall (PR) Curve**

Both require your classifier to output **class probabilities**, not just class labels.

---

## 🔷 II. ROC Curve (One-vs-Rest)

### ✅ What It Plots:

* **True Positive Rate (TPR)** = Recall = $\frac{TP}{TP + FN}$
* **False Positive Rate (FPR)** = $\frac{FP}{FP + TN}$

**ROC Curve** plots:

$$
\text{TPR vs. FPR at different thresholds}
$$

### ✅ In Multi-Class:

Use **One-vs-Rest (OvR)** strategy:

* Treat each class $i$ as positive, and all others as negative.
* Plot **one ROC curve per class**.

---

### ✅ Area Under Curve (AUC-ROC):

$$
\text{AUC}_{i} = \int_0^1 \text{TPR}_{i}(FPR) \, dFPR
$$

* $\text{AUC} \in [0, 1]$
* 1 = perfect, 0.5 = random, <0.5 = worse than random

You can compute:

* **Macro-average AUC**: average of per-class AUCs.
* **Weighted-average AUC**: weighted by class prevalence.

---

## 🔷 III. Precision-Recall Curve (One-vs-Rest)

### ✅ What It Plots:

* **Precision vs. Recall** at different thresholds

This is especially helpful when **classes are imbalanced**.

### ✅ In Multi-Class:

* One PR curve per class using OvR.
* Compute **Average Precision (AP)** per class.

---

### ✅ Average Precision (AP)

The area under the **PR curve**:

$$
\text{AP}_i = \int_0^1 \text{Precision}_i(\text{Recall}) \, dRecall
$$

Then average:

* **Macro-averaged PR AUC**: treat all classes equally.
* **Weighted PR AUC**: weighted by class support.

---

## 🔷 IV. Comparison: ROC vs. PR Curves

| Aspect                  | ROC Curve                       | PR Curve                            |
| ----------------------- | ------------------------------- | ----------------------------------- |
| Y-axis                  | TPR (Recall)                    | Precision                           |
| X-axis                  | FPR                             | Recall                              |
| Handles Imbalance Well? | ❌ No (may give optimistic view) | ✅ Yes                               |
| Best Use                | Balanced datasets               | Imbalanced datasets                 |
| Interpretation          | Class separability              | Focus on positive class performance |

---


---

## 🔷 VI. When to Use Curves in Practice

| Use Case                                               | Preferred Curve                 |
| ------------------------------------------------------ | ------------------------------- |
| Class balance is good                                  | ROC Curve                       |
| Class imbalance                                        | PR Curve                        |
| Evaluating probabilistic output (e.g., Softmax/Logits) | Both                            |
| Choosing threshold                                     | PR or ROC curve                 |
| Deployment decision making                             | PR curve shows trade-off better |
